```python
#12
top5_cases = df.nlargest(5, 'Total Cases')[['Country', 'Total Cases']]
top5_cases
```

```python
#13
plt.figure(figsize=(10, 6))
sns.barplot(data=top5_cases, x='Country', y='Total Cases', palette='Reds_r')
plt.title('Top 5 Countries with Highest COVID-19 Cases')
plt.xlabel('Country')
plt.ylabel('Total Cases')
plt.show()
```

```python
#14
lowest5_cases = df.nsmallest(5, 'Total Cases')[['Country', 'Total Cases']]
lowest5_cases
```

```python
#15
plt.figure(figsize=(10, 6))
sns.barplot(data=lowest5_cases, x='Country', y='Total Cases', palette='Blues_r')
plt.title('5 Countries with Lowest COVID-19 Cases')
plt.xlabel('Country')
plt.ylabel('Total Cases')
plt.show()
```

```python
#16
top5_deaths = df.nlargest(5, 'Total Deaths')[['Country', 'Total Deaths']]
top5_deaths
```

```python
#17
plt.figure(figsize=(10, 6))
sns.barplot(data=top5_deaths, x='Country', y='Total Deaths', palette='Oranges_r')
plt.title('Top 5 Countries with Highest COVID-19 Deaths')
plt.xlabel('Country')
plt.ylabel('Total Deaths')
plt.show()
```

```python
#18
lowest5_deaths = df.nsmallest(5, 'Total Deaths')[['Country', 'Total Deaths']]
lowest5_deaths
```

```python
#19
top5_death_pct = df.nlargest(5, 'Death percentage')[['Country', 'Death percentage']]
top5_death_pct
```

```python
#20
plt.figure(figsize=(10, 6))
sns.barplot(data=top5_death_pct, x='Country', y='Death percentage', palette='Purples_r')
plt.title('Top 5 Countries with Highest COVID-19 Death Percentage')
plt.xlabel('Country')
plt.ylabel('Death Percentage (%)')
plt.show()
```

```python
#21
continent_cases = df.groupby('Continent')['Total Cases'].sum()
continent_cases
```

```python
#22
continent_cases_df = continent_cases.reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=continent_cases_df, x='Continent', y='Total Cases', hue='Continent', palette='viridis', legend=False)
plt.title('Total COVID-19 Cases by Continent')
plt.xlabel('Continent')
plt.ylabel('Total Cases')
plt.xticks(rotation=45)
plt.show()
```

```python
#23
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Population', y='Total Cases', alpha=0.7, color='teal')
plt.title('Relationship Between Population and Total COVID-19 Cases')
plt.xlabel('Population')
plt.ylabel('Total Cases')
plt.show()
```

```python
#24
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Total Cases', y='Total Deaths', alpha=0.7, color='crimson')
plt.title('Relationship Between Total COVID-19 Cases and Total Deaths')
plt.xlabel('Total Cases')
plt.ylabel('Total Deaths')
plt.show()
```

```python
#25
# Calculate Spearman rank correlation
spearman_corr = df.corr(method='spearman', numeric_only=True)
print(spearman_corr)

# Visualize the correlation matrix with a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(spearman_corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Spearman Rank Correlation Heatmap')
plt.show()

```
